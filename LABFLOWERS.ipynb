```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('/content/IRIS.csv.xls')
```

```python
df.head()
```

```python
df.shape
```

```python
print("\nBasic Information:")
print(df.info())
print("\nSummary Statistics:")
print(df.describe())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x=df.iloc[:, -1])  # Assuming the last column is the target class
plt.title("Class Distribution")
plt.show()
```

```python
# Pairplot to visualize relationships between features
sns.pairplot(df, hue=df.columns[-1])  # Color by the target class
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df.iloc[:, :-1].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Boxplots to check for outliers
plt.figure(figsize=(12, 6))
df.iloc[:, :-1].boxplot()
plt.title("Boxplot of Features")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
df.iloc[:, :-1].boxplot()
plt.title("Boxplot of Features")
plt.show()
```

```python
# Histogram for feature distributions
df.iloc[:, :-1].hist(figsize=(12, 8), bins=15)
plt.suptitle("Feature Distributions")
plt.show()
```

```python
print("\nSkewness of Features:")
print(df.iloc[:, :-1].skew())

print("\nKurtosis of Features:")
print(df.iloc[:, :-1].kurtosis())
```

```python
plt.figure(figsize=(12, 6))
for col in df.columns[:-1]:
    sns.kdeplot(df[col], shade=True, label=col)
plt.legend()
plt.title("Kernel Density Estimation of Features")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df.iloc[:, 0], y=df.iloc[:, 1], hue=df.iloc[:, -1], palette='viridis')
plt.xlabel(df.columns[0])
plt.ylabel(df.columns[1])
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.show()
```

```python
sns.pairplot(df, hue=df.columns[-1], diag_kind='kde', palette='coolwarm')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df.iloc[:, 2], y=df.iloc[:, 3], hue=df.iloc[:, -1], palette='plasma')
plt.xlabel(df.columns[2])
plt.ylabel(df.columns[3])
plt.title("Scatter Plot of Petal Length vs Petal Width")
plt.show()

```
